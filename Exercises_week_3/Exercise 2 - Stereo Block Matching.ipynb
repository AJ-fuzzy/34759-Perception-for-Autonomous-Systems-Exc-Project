# Stereo template matching/block matching


```python
import cv2
import numpy as np
from matplotlib import pyplot as plt
from skimage import data

plt.gray()
```

Start by loading a set of corresponding left and right image. We also resize the images since the computations we expect to do can take a long time to process. Resizing is optional, but recommended atleast until you have verified your computer is capable of doing the computations within reasonable time.

```python
# load the images
img_left = cv2.imread("playtable_left.png")
img_right = cv2.imread("playtable_right.png")

# rescale images (can be left out)
img_size = (int(img_left.shape[1]/5), int(img_left.shape[0]/5))
img_left = cv2.resize(img_left, img_size, interpolation=cv2.INTER_AREA)
img_right = cv2.resize(img_right, img_size, interpolation=cv2.INTER_AREA)

# convert images to grayscale for template matching
gray_left = cv2.cvtColor(img_left, cv2.COLOR_BGR2GRAY)
gray_right = cv2.cvtColor(img_right, cv2.COLOR_BGR2GRAY)

f, (ax_left, ax_right) = plt.subplots(1, 2, figsize=(18,18))
ax_left.imshow(gray_left)
ax_right.imshow(gray_right)
```

Next step is to create a disparity map based on the two stereo images. This is done using the same procedure as we did with exercise 1, except now we look for blocks of pixel from the left image in the right image, or vice versa.

## Exercise 2.1
You will have to play with the below parameters until you find a good solution that gives a representative 3D projection. Reflect/search on what those parameters actually mean and how they influence the outcome.

`setMinDisparity`: The minimum possible disparity (pixel shift) to search for. Disparity is inversely proportional to depth. By setting it to 7 instead of 0, you tell the algorithm to ignore objects that are infinitely far away and start searching at a 7-pixel offset.

`numDisparities`: The maximum range of pixel shifts to search. This value must be divisible by 16. Combined with min_disp = 7, the algorithm will only search for matches that have shifted between 7 and 54 pixels. This defines your "depth search window."

`blockSize`: The dimensions of the sliding window (5x5 pixels) used to compare the left and right images. A smaller block size like 5 captures finer edge details but produces a noisier image. A larger block size produces smoother depth maps but blurs object edges. Must be an odd number.

`setDisp12MaxDiff`: Controls the left-right consistency check. The algorithm computes disparity from left-to-right, then right-to-left. If the difference between the two calculations exceeds 200 pixels, the pixel is rejected. A value of 200 is extremely high, which effectively disables this safety check.

`setUniquenessRatio`: The "confidence" threshold. The algorithm will only accept a match if the best match is at least 1% better than the second-best match. A ratio of 1 is very low (usually this is 5-15), meaning the algorithm will accept highly uncertain matches, resulting in a dense but noisy depth map.

`setSpeckleWindowSize` and `setSpeckleRange`: These act as a post-processing noise filter to remove tiny blobs of incorrect matches. It will delete any isolated blob of pixels that is smaller than 3 pixels (SpeckleWindowSize) where the disparity within that blob varies by less than 3 (SpeckleRange).

```python
min_disp = 7
num_disp = 3 * 16
block_size = 5
stereo = cv2.StereoBM_create(numDisparities = num_disp, blockSize = block_size)
stereo.setMinDisparity(min_disp)
stereo.setDisp12MaxDiff(200)
stereo.setUniquenessRatio(1)
stereo.setSpeckleRange(3)
stereo.setSpeckleWindowSize(3)
disp = stereo.compute(gray_left, gray_right).astype(np.float32) / 16.0

min_disp = 7
num_disp = 3 * 16
block_size = 7
stereo = cv2.StereoBM_create(numDisparities = num_disp, blockSize = block_size)
stereo.setMinDisparity(min_disp)
stereo.setDisp12MaxDiff(50)
stereo.setUniquenessRatio(4)
stereo.setSpeckleRange(5)
stereo.setSpeckleWindowSize(3)

disp_n = stereo.compute(gray_left, gray_right).astype(np.float32) / 16.0

plt.figure(figsize=(36,36))
plt.subplot(1,2,1)
plt.imshow(disp)
plt.subplot(1,2,2)
plt.imshow(disp_n)
```

## Exercise 2.2
Once you have a disparity map without too much noise, you can use the below to export you disparity map to a .ply file that you can load and display in a 3D visualization software (e.g. [Meshlab](https://www.meshlab.net/#description)).

**Hint:** To open meshlab type `meshlab` in a terminal. 
To vizualize the point cloud, drag the generated `out.ply` into the meshlab scene or click `File` > `Import Mesh` and navigate to where our source-code is located and open `out.ply`. 

```python
def export_pointcloud(disparity_map, colors):
    ply_header = '''ply
    format ascii 1.0
    element vertex %(vert_num)d
    property float x
    property float y
    property float z
    property uchar red
    property uchar green
    property uchar blue
    end_header
    '''

    def write_ply(fn, verts, colors):
        verts = verts.reshape(-1, 3)
        colors = colors.reshape(-1, 3)
        verts = np.hstack([verts, colors])
        with open(fn, 'wb') as f:
            f.write((ply_header % dict(vert_num=len(verts))).encode('utf-8'))
            np.savetxt(f, verts, fmt='%f %f %f %d %d %d ')

    h, w = disparity_map.shape[:2]
    f = .9 * w  # guess for focal length. If you 3D reconstruction looks skewed in the viewing direction, try adjusting this parameter.
    Q = np.float32([[1, 0, 0, -0.5 * w],
                    [0, -1, 0, 0.5 * h],  # turn points 180 deg around x-axis,
                    [0, 0, 0, -f],  # so that y-axis looks up
                    [0, 0, 1, 0]])
    points = cv2.reprojectImageTo3D(disparity_map, Q)
    
    mask = disparity_map > disparity_map.min()
    out_points = points[mask]
    out_colors = colors[mask]
    out_fn = 'out.ply'
    write_ply('out.ply', out_points, out_colors)
    print(f'{out_fn} saved')

export_pointcloud(disparity_map=disp, colors=img_left)
```
